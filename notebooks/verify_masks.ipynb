```python
from pathlib import Path 

import SimpleITK as sitk 

import numpy as np 



dataset = ["rescale"]

path_base = Path("/projects/vig/Datasets/aneurysm/cta_datasets/")


path_base = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane")

stats_header = ["case_name", "num_voxels_art", "num_voxels_ven", "num_voxels_brain", "num_voxels_cvs"]
stats = []
import tqdm 
for ds in dataset:
    path_ds = path_base / ds 
    path_vessel = path_ds / "crop_0.4_vessel"
    path_brain = path_ds / "crop_0.4_totalseg"
    path_cvs_mask = path_ds / "cvs_mask"
    for case in tqdm.tqdm(path_vessel.iterdir()):
        header = sitk.ReadImage(str(case))
        vessel = sitk.GetArrayFromImage(header)
        num_voxels_art = np.sum(vessel == 1)
        num_voxels_ven = np.sum(vessel == 2)
        case_name = str(case.name).replace(".nii.gz", "")
        header_brain = sitk.ReadImage(str(path_brain / f"{case_name}/brain.nii.gz"))
        brain = sitk.GetArrayFromImage(header_brain)
        num_voxels_brain = np.sum(brain > 0)
        try:
            header_cvs = sitk.ReadImage(str(path_cvs_mask / f"{case_name}.nii.gz"))
            cvs_mask = sitk.GetArrayFromImage(header_cvs)
            num_voxels_cvs = np.sum(cvs_mask > 0)
        except:
            num_voxels_cvs = -1
        stats.append([case_name, num_voxels_art, num_voxels_ven, num_voxels_brain, num_voxels_cvs])

```

```python
# print as a table with justification etc

for col in stats_header:
    print(f"{col:25s}", end="")
print()
for row in (stats):
    for i, item in enumerate(row):
        if i==0:
            print(f"{str(item):75s}", end=" | ")
        else:
            print(f"{str(item):25s}", end="")
    print()
    
```

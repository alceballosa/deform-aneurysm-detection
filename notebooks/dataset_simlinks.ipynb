```python
from pathlib import Path
import os
import shutil
import pandas as pd

path_ds_1 = "/projects/vig/Datasets/aneurysm/cta_datasets/internal_train"

path_ds_2 = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane"

target_ds = "/projects/vig/Datasets/aneurysm/cta_datasets/merged_train"

relevant_folders = ["crop_0.4", "crop_0.4_vessel_edt_comp"]

labels_ds_1 = pd.read_csv(
    "/projects/vig/alberto/medical/deform-aneurysm-detection/labels/gt/internal_train_crop_0.4_mixed_fps_fix.csv"
)

labels_ds_2 = pd.read_csv(
    "/projects/vig/alberto/medical/deform-aneurysm-detection/labels/gt/source_files_rsna/annotations_final.csv"
)
# merge label files and save
labels_target_path = (
    "/projects/vig/Datasets/aneurysm/cta_datasets/merged_train/annotations.csv"
)

df_merged = pd.concat([labels_ds_1, labels_ds_2], ignore_index=True)
df_merged.to_csv(labels_target_path, index=False)
df_merged.to_csv(labels_target_path, index=False)

#labels_ds_1.columns
```

```python
# for every relevant folder, create symlinks for every file in both datasets
for folder in relevant_folders:
    folder_path_1 = os.path.join(path_ds_1, folder)
    folder_path_2 = os.path.join(path_ds_2, folder)
    target_folder_path = os.path.join(target_ds, folder)
    # copy labels

    # dataset 1
    for file_name in os.listdir(folder_path_1):
        source_file = os.path.join(folder_path_1, file_name)
        target_file = os.path.join(target_folder_path, file_name)
        if not os.path.exists(target_file):
            os.symlink(source_file, target_file)

    # dataset 2
    for file_name in os.listdir(folder_path_2):
        source_file = os.path.join(folder_path_2, file_name)
        target_file = os.path.join(target_folder_path, file_name)
        if not os.path.exists(target_file):
            os.symlink(source_file, target_file)
```

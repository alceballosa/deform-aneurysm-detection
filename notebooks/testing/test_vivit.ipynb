```python
import torch
import sys
import einops

sys.path.append("../../src")

from models.deformable.vivit import ViT

v = ViT(
    image_size=128,  # image size
    frames=64,  # number of frames
    image_patch_size=4,  # image patch size
    frame_patch_size=4,  # frame patch size
    num_classes=1000,
    dim=1024,
    spatial_depth=6,  # depth of the spatial transformer
    temporal_depth=6,  # depth of the temporal transformer
    heads=8,
    mlp_dim=2048,
    variant="factorized_self_attention",  # or 'factorized_self_attention'
)

video = torch.randn(4, 3, 64, 64, 64)  # (batch, channels, frames, height, width)

preds = v(video)  # (4, 1000)
preds.shape
preds = preds[:, :, 1:, :]
spatial_dim = int(preds.shape[-2] ** 0.5)
```

```python
import einops
```

```python
# shape is b z (x y) c, turn into b z x y
preds = einops.rearrange(
    preds, "b z (x y) c -> b z x y c", x=spatial_dim, y=spatial_dim
)
preds.shape
```

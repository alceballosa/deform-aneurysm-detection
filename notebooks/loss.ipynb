```python
from pathlib import Path 

path_crop = "/data/pt_sinoct/crop_0.4_vessel"
path_crop = Path(path_crop)
# get all filenames 
filenames = [str(p.name) for p in path_crop.glob("*.nii.gz")]
len(filenames)
```

```python
path_og = Path("/data/pt_sinoct/crop_0.4")

filenames_og = [str(p.name) for p in Path(path_og).glob("*.nii.gz")]

len(filenames_og)

intersection = set(filenames).intersection(set(filenames_og))

len(intersection)
```

```python
# move all files in  intersection from crop_0.4 to crop_0.4_done 

import shutil 

path_crop_done = Path("/data/pt_sinoct/crop_0.4_done")

for f in intersection:
    shutil.move(str(path_og / f), str(path_crop_done / f))
```

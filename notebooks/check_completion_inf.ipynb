```python
from pathlib import Path

import os

path_results = "../results"
datasets = ["hospital140", "cmha", "internal_test", "external"]

model_keys = ["nov", "done", "merged"]

path_weights = "../model_weights"

# get all models across folder fititng the key

list_models = []
for model_key in model_keys:
    models_key = [
        d.name
        for d in Path(path_weights).iterdir()
        if model_key in d.name and d.is_dir()
    ]
    list_models += models_key

list_models = sorted(list(set(list_models)))
list_models
```

```python
infs_short = ["16k", "final"]
infs_merged = ["26k", "30k", "final"]

total = 0
total_missed = 0
for i, dataset in enumerate(datasets):
    for model in list_models:
        if "short" in model:
            infs = infs_short
        elif "merged" in model:
            infs = infs_merged
        else:
            continue
        path_ds = os.path.join(path_results, dataset, model)
        for inf in infs:
            path_inf = os.path.join(path_ds, "inference_" + inf, "predict.csv")
            if not os.path.exists(path_inf):
                print(f"Missing: {path_inf}")
                total_missed += 1
            total += 1
```

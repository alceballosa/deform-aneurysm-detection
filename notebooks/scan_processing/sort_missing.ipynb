```python
from pathlib import Path

path_scans = "/data/pt_rsna/og"

path_vessels = "/data/pt_rsna/og_0.4"

path_target = "/data/pt_rsna/og_done"




# get names of missing files in vessels

scans = [f.name for f in Path(path_scans).iterdir()]
og = [f.name for f in Path(path_vessels).iterdir()]

# intersection
done = set(scans).intersection(og)

# move all in intersection to target

for f in Path(path_scans).iterdir():
    if f.name in done:
        f.rename(Path(path_target) / f.name)
```

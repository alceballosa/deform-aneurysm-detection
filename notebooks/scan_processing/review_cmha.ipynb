```python
import SimpleITK as sitk

from pathlib import Path

path_cmha_og = "/data/aneurysm/cmha/og"
path_cmha_annot = "/data/aneurysm/cmha/crop_0.4_label"

# get lists of file from both 

files_og = sorted(Path(path_cmha_og).glob("*.nii.gz"))
files_annot = sorted(Path(path_cmha_annot).glob("*.nii.gz"))

files_og = [Path(f).name for f in files_og]
files_annot = [Path(f).name for f in files_annot]

# check if the number of files is the same

if len(files_og) != len(files_annot):
    print("Number of files in og and annot directories do not match")
    exit(1)
```

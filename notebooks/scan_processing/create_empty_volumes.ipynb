```python
from pathlib import Path 
import SimpleITK as sitk
import numpy as np


path_controls= Path("/data/cmha/controls")
path_target = Path("/projects/vig/Datasets/aneurysm/cta_datasets/cmha/og_label")

for control in path_controls.iterdir():
    niigz_file = list(control.glob("*.nii.gz"))[0]
    header = sitk.ReadImage(str(niigz_file))
    array = sitk.GetArrayFromImage(header)

    # create a new array with the same shape but empty -just 0s-
    new_array = np.zeros_like(array)

    # save 
    new_header = sitk.GetImageFromArray(new_array)
    new_header.CopyInformation(header)
    sitk.WriteImage(new_header, str(path_target / niigz_file.name ))




```

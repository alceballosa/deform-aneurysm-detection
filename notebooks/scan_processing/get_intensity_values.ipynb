```python
import SimpleITK as sitk 
import numpy as np 
from pathlib import Path 
import tqdm
```

```python
dataset = "internal_train" 
path_dataset = Path(f"/projects/vig/Datasets/aneurysm/cta_datasets/{dataset}/crop_0.4")
path_label = Path(f"/projects/vig/Datasets/aneurysm/cta_datasets/{dataset}/crop_0.4_label")
path_vessel = Path(f"/projects/vig/Datasets/aneurysm/cta_datasets/{dataset}/crop_0.4_vessel")

scan_paths = sorted(list(path_dataset.glob("*.nii.gz")))
label_paths = sorted(list(path_label.glob("*.nii.gz")))
vessel_paths = sorted(list(path_vessel.glob("*.nii.gz")))

all_intensity_hu_means = []
all_aneurysm_hu_values = []
all_vessel_hu_values = []


for i, _ in tqdm.tqdm(enumerate(scan_paths)):
    scan_path = scan_paths[i]
    label_path = label_paths[i]
    scan = sitk.GetArrayFromImage(sitk.ReadImage(scan_path))
    label = sitk.GetArrayFromImage(sitk.ReadImage(label_path))
    vessel = sitk.GetArrayFromImage(sitk.ReadImage(vessel_paths[i]))
    aneurysm_mask = label == 1 
    vessel_mask = vessel > 0
    intensity_hu_mean = scan.mean()
    aneurysm_hu_values = scan[aneurysm_mask]
    all_aneurysm_hu_values.extend(aneurysm_hu_values)
    vessel_hu_values = scan[vessel_mask]
    all_vessel_hu_values.extend(vessel_hu_values)
    all_intensity_hu_means.append(intensity_hu_mean)
    



    
```

```python
# get the 5th and 95th percentiles of all arrays

all_intensity_hu_means = np.array(all_intensity_hu_means)
all_aneurysm_hu_values = np.array(all_aneurysm_hu_values)
all_vessel_hu_values = np.array(all_vessel_hu_values)   
intensity_hu_mean_5th = np.percentile(all_intensity_hu_means, 5)
intensity_hu_mean_95th = np.percentile(all_intensity_hu_means, 95)
aneurysm_hu_values_5th = np.percentile(all_aneurysm_hu_values, 5)
aneurysm_hu_values_95th = np.percentile(all_aneurysm_hu_values, 95)
vessel_hu_values_5th = np.percentile(all_vessel_hu_values, 5)
vessel_hu_values_95th = np.percentile(all_vessel_hu_values, 95)
print(f"Intensity HU mean 5th percentile: {intensity_hu_mean_5th}")
print(f"Intensity HU mean 95th percentile: {intensity_hu_mean_95th}")
print(f"Aneurysm HU values 5th percentile: {aneurysm_hu_values_5th}")
print(f"Aneurysm HU values 95th percentile: {aneurysm_hu_values_95th}")
print(f"Vessel HU values 5th percentile: {vessel_hu_values_5th}")
print(f"Vessel HU values 95th percentile: {vessel_hu_values_95th}")
```

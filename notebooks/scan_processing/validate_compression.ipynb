```python
path_vessel = "/projects/vig/Datasets/aneurysm/cta_datasets/internal_train/crop_0.4_vessel"

path_compressed = "/projects/vig/Datasets/aneurysm/cta_datasets/internal_train/crop_0.4_compressed"


```

```python
import SimpleITK as sitk 
from pathlib import Path 
import tqdm 

# for each scan, load both the vessel and compressed, then check how many vessel voxels (1 or 2) are == -512
#

vessel_files = sorted(Path(path_vessel).glob("*.nii.gz"))
compressed_files = sorted(Path(path_compressed).glob("*.nii.gz"))

counts_512 = []

for vessel_f, compressed_f in tqdm.tqdm(zip(vessel_files, compressed_files), total=len(vessel_files)):
    vessel_img = sitk.ReadImage(str(vessel_f))
    vessel_arr = sitk.GetArrayFromImage(vessel_img)

    compressed_img = sitk.ReadImage(str(compressed_f))
    compressed_arr = sitk.GetArrayFromImage(compressed_img)

    # check how many vessel voxels are == -512 in the compressed
    vessel_voxels = (vessel_arr > 0)  # assuming vessel voxels are > 0
    count_512 = ((compressed_arr == -512) & vessel_voxels).sum()
    counts_512.append(count_512)
    if count_512 > 0:
        print(f"{vessel_f.name}: {count_512} vessel voxels are -512 in compressed")
```

```python
path_vessel = "/projects/vig/Datasets/aneurysm/cta_datasets/internal_train/crop_0.4_vessel"

path_uncompressed = "/projects/vig/Datasets/aneurysm/cta_datasets/internal_train/crop_0.4"

# for each scan, load both the vessel and uncompressed, then check how many vessel voxels (1 or 2) are <= -512
vessel_files = sorted(Path(path_vessel).glob("*.nii.gz"))
uncompressed_files = sorted(Path(path_uncompressed).glob("*.nii.gz"))
counts_512_uncompressed = []
for vessel_f, uncompressed_f in tqdm.tqdm(zip(vessel_files, uncompressed_files), total=len(vessel_files)):
    vessel_img = sitk.ReadImage(str(vessel_f))
    vessel_arr = sitk.GetArrayFromImage(vessel_img)

    uncompressed_img = sitk.ReadImage(str(uncompressed_f))
    uncompressed_arr = sitk.GetArrayFromImage(uncompressed_img)

    # check how many vessel voxels are <= -512 in the uncompressed
    vessel_voxels = (vessel_arr > 0)  # assuming vessel voxels are > 0
    count_512_uncompressed = ((uncompressed_arr <= -512) & vessel_voxels).sum()
    counts_512_uncompressed.append(count_512_uncompressed)
    if count_512_uncompressed > 0:
        print(f"{vessel_f.name}: {count_512_uncompressed} vessel voxels are <= -512 in uncompressed")


```

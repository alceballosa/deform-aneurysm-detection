```python
from pathlib import Path
import SimpleITK as sitk 
import numpy as np
import tqdm

path_vessel = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4" 
path_vessel = Path(path_vessel)
files = list(path_vessel.glob("*.nii.gz"))

dict_data = {
    "filename": [],
    "mean": [],
    "percentile_25": [],
    "percentile_75": [],
    "max": [],
    "min": [],
}

for file in tqdm.tqdm(files):
    try:
        header = sitk.ReadImage(file)
        vol     = sitk.GetArrayFromImage(header)
        dict_data["filename"].append(file.name)
        dict_data["mean"].append(np.mean(vol))
        dict_data["percentile_25"].append(np.percentile(vol, 25))
        dict_data["percentile_75"].append(np.percentile(vol, 75))
        dict_data["max"].append(np.max(vol))
        dict_data["min"].append(np.min(vol))
    except:
        print(f"Error reading file {file}")
        continue


# write to json file 


```

```python
import SimpleITK as sitk 
from pathlib import Path 
import json

path_nifti = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/rsna/nifti")
path_label_nifti = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/rsna_intermediate/nifti_label")

dest_dir = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna/og")
dest_dir_label = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna/og_label")

usable_niftis = []

all_label_files = list(path_label_nifti.iterdir())
for file in all_label_files:
    if file.suffix == ".gz":
        nifti_name = file.name
        label_itk = sitk.ReadImage(str(file))
        label_array = sitk.GetArrayFromImage(label_itk)
        # check the minimum shape of the array
        min_shape = min(label_array.shape)
        if min_shape >= 95:
            nifti_name = file.name 
            usable_niftis.append(nifti_name)
             

usable_niftis
```

```python
import json 
template_labels = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/rsna/nifti_label/{name}"
template_niftis = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/rsna/nifti/{name}"
usable_ct_niftis = []

for name in usable_niftis:
    try:
        label_path = template_labels.format(name=name)
        img_path = template_niftis.format(name=name)
        json_path = img_path.replace(".nii.gz", ".json")
        dest_path = str(dest_dir / name)
        dest_path_label = str(dest_dir_label / name)

        with open(json_path, 'r') as f:
            json_dict = json.load(f)
        
        if json_dict["Modality"] == "CT":
            usable_ct_niftis.append(name)
    except: 
        print(f"Skipping {name}")
```

```python
import os 

# create dirs
dest_dir.mkdir(parents=True, exist_ok=True)
dest_dir_label.mkdir(parents=True, exist_ok=True)

for name in usable_ct_niftis:
    label_path = template_labels.format(name=name)
    img_path = template_niftis.format(name=name)
    dest_path = str(dest_dir / name)
    dest_path_label = str(dest_dir_label / name)

    os.system(f"cp {img_path} {dest_path}")
    os.system(f"cp {label_path} {dest_path_label}")
```

```python
from pathlib import Path


path_split = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4_split"
path_split = Path(path_split)

# get all subfolders and subsubfiles 

scans = []
folders = []
for subfolder in path_split.iterdir():
    if subfolder.is_dir():
        for file in subfolder.iterdir():
            scans.append(file.name)
            folders.append(subfolder.name)

path_seg = "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4_vessel"
path_seg = Path(path_seg)

seg_scans = []
for file in path_seg.iterdir():
    seg_scans.append(file.name)



```

```python
is_missing_from_seg = []

for scan, folder in zip(scans, folders):
    if scan not in seg_scans:
        is_missing_from_seg.append((scan, folder))
is_missing_from_seg, len(is_missing_from_seg)
```

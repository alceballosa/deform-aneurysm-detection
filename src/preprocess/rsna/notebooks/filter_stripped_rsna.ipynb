```python
import SimpleITK as sitk
import numpy as np
from pathlib import Path
import tqdm

path = Path(
    "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4_totalseg"
)
counter = 0
filt_files = []
for fil in tqdm.tqdm(path.iterdir()):
    file_path = fil / "brain.nii.gz"
    header = sitk.ReadImage(str(file_path))
    array = sitk.GetArrayFromImage(header)
    suma = np.sum(array)
    if suma < 1000000:
        print(fil.name, suma)
        filt_files.append(fil.name)
        counter += 1
```

```python
path_crop = Path("/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4")
path_crop_backup_stripped = Path( "/scratch/ceballosarroyo.a/aneurysm/mm_datasets/cta_rsna_ane/crop_0.4_stripped")

# move all files in filt_files from path_crop to path_crop_backup_stripped
import shutil
import os
path_crop_backup_stripped.mkdir(parents=True, exist_ok=True)
for fil_name in filt_files:
    shutil.move(path_crop / (fil_name+".nii.gz"), path_crop_backup_stripped / (fil_name + ".nii.gz"))
```
